# 02. Shape Anisotropy and Separatrix Aggregation

Investigating how particle eccentricity $lpha$ breaks phase space volume conservation (Liouville theorem) and drives clustering near vortex separatrices.


```python
import sys, os
sys.path.insert(0, os.path.abspath("../src"))

import numpy as np
import matplotlib.pyplot as plt
from active_matter import simulate_ensemble, plot_particles_2d, eccentricity_from_aspect_ratio, set_visual_style
```

### 1. Shape Parameter (Bretherton / Jeffrey Alignment)
For an ellipsoid with major axis $ and minor axis $, the aspect ratio is $\gamma = a/b$.
The alignment parameter is $lpha = rac{\gamma^2 - 1}{\gamma^2 + 1}$.


```python
# Compute alignment alpha for a sphere (gamma=1) and rod (gamma=10)
gammas = [1.0, 1.5, 3.0, 10.0]
for g in gammas:
    print(f"Aspect ratio gamma = {g:.1f} -> Eccentricity alpha = {eccentricity_from_aspect_ratio(g):.4f}")
```

### 2. Multi-Particle Simulation: Spheres vs Ellipsoids
Simulating 2000 particles to observe macroscopic spatial distributions.


```python
np.random.seed(42)
N = 2000
L = 2.0 * np.pi
X0 = np.random.uniform(0.0, L, size=(N, 3))

# Spherical particles (alpha = 0.0): Phase space divergence is zero
traj_spheres, _ = simulate_ensemble(X0, dt=0.01, nt=2500, nsave=2500, vp=0.05, alpha=0.0, wrap_pbc=True)
plot_particles_2d(traj_spheres[-1], title="Spheres (alpha = 0.0): Uniform Phase Space Distribution")

# Slender ellipsoids (alpha = 0.99): Phase space contraction causes separatrix aggregation
traj_rods, _ = simulate_ensemble(X0, dt=0.01, nt=2500, nsave=2500, vp=0.05, alpha=0.99, wrap_pbc=True)
plot_particles_2d(traj_rods[-1], title="Ellipsoids (alpha = 0.99): Separatrix Aggregation")
```
