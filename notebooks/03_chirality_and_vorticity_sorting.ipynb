# 03. Chirality and Vorticity-Based Micro-Sorting

Microorganisms like *E. coli* or spermatozoa often exhibit built-in rotational chirality ( 
eq 0$). This notebook demonstrates how chirality breaks parity and enables passive sorting in vortex arrays.


```python
import sys, os
sys.path.insert(0, os.path.abspath("../src"))

import numpy as np
import matplotlib.pyplot as plt
from active_matter import simulate_ensemble, compute_vorticity, compute_vorticity_pdf_ratio, set_visual_style
```

### 1. Simulating Chiral vs Non-Chiral Ensembles


```python
np.random.seed(42)
N = 4000
L = 2.0 * np.pi
X0 = np.random.uniform(0.0, L, size=(N, 3))

# Baseline uniform spatial sample
X_unif = np.random.uniform(0.0, L, size=(N, 2))
w_unif = compute_vorticity(X_unif[:, 0], X_unif[:, 1], L=L)

# Symmetrical non-chiral particles (C = 0.0)
traj_nc, _ = simulate_ensemble(X0, dt=0.01, nt=3000, nsave=3000, vp=0.15, alpha=0.99, chirality=0.0, wrap_pbc=True)
w_nc = compute_vorticity(traj_nc[-1, :, 0], traj_nc[-1, :, 1], L=L)

# Chiral particles with clockwise/counter-clockwise turning (C = 0.3)
traj_c, _ = simulate_ensemble(X0, dt=0.01, nt=3000, nsave=3000, vp=0.15, alpha=0.99, chirality=0.3, wrap_pbc=True)
w_c = compute_vorticity(traj_c[-1, :, 0], traj_c[-1, :, 1], L=L)
```

### 2. Computing the Vorticity Sampling Ratio (\omega) / P_0(\omega)$


```python
bins_nc, r_nc, _, _ = compute_vorticity_pdf_ratio(w_nc, w_unif)
bins_c, r_c, _, _ = compute_vorticity_pdf_ratio(w_c, w_unif)

set_visual_style()
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(bins_nc, r_nc, label="Non-Chiral ( = 0.0$)", color="#333333", linewidth=2)
ax.plot(bins_c, r_c, label="Chiral ( = 0.3$)", color="#e63946", linewidth=2)
ax.set_xlabel(r"Vorticity ($\omega$)")
ax.set_ylabel(r"Sampling Ratio (\omega) / P_0(\omega)$")
ax.set_title("Chirality-Induced Vorticity Selection")
ax.grid(True, linestyle=":", alpha=0.6)
ax.legend(); plt.show()
```
