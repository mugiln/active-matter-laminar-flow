# 01. Cellular Flow and Microswimmer Trapping

This notebook demonstrates 2D laminar cellular vortex flows and contrasts passive tracers with self-propelled microswimmers.


```python
import sys, os
sys.path.insert(0, os.path.abspath("../src"))

import numpy as np
import matplotlib.pyplot as plt
from active_matter import cellular_velocity, compute_vorticity, simulate_ensemble, set_visual_style
```

### 1. Visualizing the Cellular Flow Field
The flow is derived from streamfunction $\psi(x, y) = rac{U_0 L}{2\pi} \sin(x) \sin(y)$.


```python
# Generate spatial grid across domain [0, 2pi]
L = 2.0 * np.pi
x = np.linspace(0, L, 25)
y = np.linspace(0, L, 25)
X_grid, Y_grid = np.meshgrid(x, y)

# Compute velocity vector components
vel = cellular_velocity(X_grid.ravel(), Y_grid.ravel(), U0=1.0, L=L)
u = vel[:, 0].reshape(25, 25)
v = vel[:, 1].reshape(25, 25)

set_visual_style()
fig, ax = plt.subplots(figsize=(6, 6))
ax.quiver(X_grid, Y_grid, u, v, color="#1d3557", alpha=0.85)
ax.set_title("2D Cellular Flow Array", fontsize=12)
ax.set_xlabel("x"); ax.set_ylabel("y")
plt.show()
```

### 2. Passive Tracer vs Active Swimmer Dynamics
Passive tracers ( = 0$) trace closed streamlines, while active swimmers ( > 0$) couple advection with self-propulsion.


```python
# Initial single particle at (1.0, 1.0) with orientation theta = pi/4
X0 = np.array([[1.0, 1.0, np.pi / 4.0]])

# Evolve passive tracer (vp = 0.0)
traj_passive, _ = simulate_ensemble(X0, dt=0.01, nt=3000, nsave=1, vp=0.0, alpha=0.0)

# Evolve active microswimmer (vp = 0.5)
traj_active, _ = simulate_ensemble(X0, dt=0.01, nt=3000, nsave=1, vp=0.5, alpha=0.0)

# Plot trajectory comparison
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(traj_passive[:, 0, 0], traj_passive[:, 0, 1], label="Passive Tracer ( = 0$)", color="#457b9d", linewidth=2)
ax.plot(traj_active[:, 0, 0], traj_active[:, 0, 1], label="Active Swimmer ( = 0.5$)", color="#e63946", linewidth=1.5)
ax.set_xlim(0, L); ax.set_ylim(0, L)
ax.set_title("Trajectory Comparison: Passive vs Active Swimmer")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.legend(); plt.show()
```
